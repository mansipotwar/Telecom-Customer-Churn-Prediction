## 6. Churn Analysis by Contract Type

### Business Question

Does the type of customer contract affect churn?

We will compare churn rates across different contract types to identify whether customers with short-term or long-term contracts are more likely to churn.

```python
# ==============================
# Contract Type vs Churn
# ==============================

# Count customers by contract type and churn status
contract_churn_count = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

# Display the result
contract_churn_count
```

```python
# ==============================
# Churn Rate by Contract Type
# ==============================

# Calculate churn percentage within each contract type
contract_churn_rate = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

# Round values to two decimal places
contract_churn_rate.round(2)
```

```python
# ==============================
# Visualize Churn Rate by Contract Type
# ==============================

# Create a bar chart showing churn rate by contract type
contract_churn_rate["Yes"].sort_values(ascending=False).plot(
    kind="bar"
)

# Add chart title and labels
plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")

# Rotate x-axis labels for readability
plt.xticks(rotation=0)

# Display the chart
plt.show()
```

💼 Business Interpretation

This suggests that long-term contracts are strongly associated with customer retention. Customers with month-to-month contracts have more flexibility to leave, while one-year and two-year contracts create greater commitment and stability.

🎯 Business Recommendation

The company could focus on converting high-risk month-to-month customers into long-term contracts by offering incentives such as discounts, loyalty benefits, or customized long-term plans.

## 7. Churn Analysis by Customer Tenure

### Business Question

Does customer tenure affect the likelihood of churn?

We will divide customers into tenure groups and compare the churn rate across different stages of the customer lifecycle.

```python
# ==============================
# Create Tenure Groups
# ==============================

# Divide customers into different tenure groups
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=[
        "0-12 Months",
        "13-24 Months",
        "25-48 Months",
        "49-72 Months"
    ]
)

# Display the first few tenure groups
df[["tenure", "TenureGroup"]].head()
```

```python
# ==============================
# Tenure Group vs Churn Count
# ==============================

# Count customers by tenure group and churn status
tenure_churn_count = pd.crosstab(
    df["TenureGroup"],
    df["Churn"]
)

# Display the result
tenure_churn_count
```

```python
# ==============================
# Calculate Churn Rate by Tenure Group
# ==============================

# Calculate churn percentage within each tenure group
tenure_churn_rate = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100

# Round the percentages to two decimal places
tenure_churn_rate.round(2)
```

```python
# ==============================
# Visualize Churn Rate by Tenure
# ==============================

# Plot the churn rate for each tenure group
tenure_churn_rate["Yes"].plot(
    kind="bar"
)

# Add chart title and labels
plt.title("Churn Rate by Customer Tenure")
plt.xlabel("Tenure Group")
plt.ylabel("Churn Rate (%)")

# Rotate labels for readability
plt.xticks(rotation=0)

# Display the chart
plt.show()
```

Key Insight

New customers (0–12 months) have the highest churn rate: 47.44%.

Matlab almost har 2 new customers mein se 1 churn kar raha hai.

As customer tenure increases, churn rate consistently decreases:

0–12 months   → 47.44%
13–24 months  → 28.71%
25–48 months  → 20.39%
49–72 months  →  9.51%
💼 Business Interpretation

The first year is the highest-risk period in the customer lifecycle. Customers who survive the initial period become significantly more loyal over time.

🎯 Business Recommendation

The company should focus heavily on new-customer onboarding and first-year retention strategies, such as early engagement, proactive support, personalized offers, and regular feedback collection.

## 10. Multivariate Analysis: Contract Type and Internet Service

```python
# ==========================================
# Churn Rate by Contract and Internet Service
# ==========================================

# Calculate churn rate for each combination of contract type and internet service
contract_internet_churn = pd.crosstab(
    [df["Contract"], df["InternetService"]],
    df["Churn"],
    normalize="index"
) * 100

# Display the churn rates
contract_internet_churn.round(2)
```

```python
# ==========================================
# Visualize Churn Rate Using a Heatmap
# ==========================================

# Extract only the churned customer percentage
churn_heatmap_data = contract_internet_churn["Yes"].unstack()

# Create the heatmap
sns.heatmap(
    churn_heatmap_data,
    annot=True,
    fmt=".2f"
)

# Add chart title
plt.title("Churn Rate by Contract Type and Internet Service")

# Add axis labels
plt.xlabel("Internet Service")
plt.ylabel("Contract Type")

# Display the chart
plt.show()
```

Strongest Finding

## 16. High-Risk Customer Profile Analysis

### Business Question

What is the churn rate among customers who have multiple high-risk characteristics?

Based on our previous analysis, we will investigate customers who:

- Have a month-to-month contract
- Use fiber optic internet service
- Do not have technical support
- Use electronic check as their payment method

This analysis will help identify a specific high-risk customer segment.

```python
# ==========================================
# Identify High-Risk Customer Segment
# ==========================================

# Filter customers with multiple high-risk characteristics
high_risk_customers = df[
    (df["Contract"] == "Month-to-month") &
    (df["InternetService"] == "Fiber optic") &
    (df["TechSupport"] == "No") &
    (df["PaymentMethod"] == "Electronic check")
]

# Display the number of customers in this segment
len(high_risk_customers)
```

```python
# ==========================================
# Churn Distribution of High-Risk Segment
# ==========================================

# Count churned and retained customers
high_risk_churn_count = high_risk_customers["Churn"].value_counts()

# Display the result
high_risk_churn_count
```

```python
# ==========================================
# Calculate High-Risk Segment Churn Rate
# ==========================================

# Calculate the percentage of customers who churned
high_risk_churn_rate = (
    high_risk_customers["Churn"]
    .value_counts(normalize=True) * 100
)

# Display rounded percentages
high_risk_churn_rate.round(2)
```

```python
# ==========================================
# Visualize High-Risk Customer Churn
# ==========================================

# Create a count plot for the high-risk segment
sns.countplot(
    data=high_risk_customers,
    x="Churn"
)

# Add chart title and labels
plt.title("Churn Distribution Among High-Risk Customers")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

# Display the chart
plt.show()
```

💼 Strong Business Insight

Customers with a month-to-month contract, fiber optic internet service, no technical support, and electronic check payment method represent a high-risk customer segment, with a churn rate of 62.92%.

🎯 Business Recommendation

The company should prioritize this segment for targeted retention campaigns. These customers could be offered long-term contract incentives, proactive technical support, personalized pricing, and alternative payment options.
